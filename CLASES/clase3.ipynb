LISTAS

```python
lista1 = [12, 12.40, "Hola", 15, 40.50,"Python"]
lista2 = list("123456789")
print(lista2)
print(type(lista2))
print(len(lista2))
print()
print(lista1)
print(type(lista1))
print(len(lista1))
```

```python
lista1 = [12, 12.40, "Hola", 15, 40.50,"Python"]
print(lista1[-1])
print(lista1[2])
print(lista1[-2])
print(lista1[4])

# modificar elementos de una lista
elemento = "Hola Python"
lista1[-1] = elemento
print(lista1)
```

Método append()

```python
colores = ["Azul", "Verde", "Amarillo", "Negro"]
print(colores)
nuevo_color = "Marron"
colores.append(nuevo_color)
color2 = "Rosado"
colores.append(color2) # agregar elementos al final
color3 = "Rojo"
colores.insert(0,color3) # agregar elementos por posición espec+ifica
print(colores)
```

```python
colores = ["Azul", "Verde", "Amarillo", "Negro"]
for color in colores:
    print(color)
```

TUPLAS

```python
miTupla = (1, "Gato", 30.40, ["Delfin", "Tigre"])
miTupla1 = 1,2,3,4,5,6
print(miTupla)
print(type(miTupla))
print(miTupla1)
print(type(miTupla1))

print(miTupla[-1])
# No se puede modificar los elementos


miTupla[-1][0] = "Leon"
print(miTupla)
```

```python
miTupla = (1, "Gato", 30.40, ["Delfin", "Tigre"])
for i in miTupla:
    print(i)
```
