Método append()

```python
colores = ["Azul", "Verde", "Amarillo", "Negro"]
print(colores)
nuevo_color = "Marron"
colores.append(nuevo_color)
color2 = "Rosado"
colores.append(color2) # agregar elementos al final
color3 = "Rojo"
colores.insert(0,color3) # agregar elementos por posición espec+ifica
print(colores)
```

EJEMPLO DE TRY - EXCEPT

```python
try:
    num1 = int(input("Ingrese un numero entero: "))
    if num1 > 10:
        print("Es mayor a 10")
    else:
        print("Es menor a 10")
except ValueError:
    print("Error en el ingreso de datos")
```

sort()

```python
lista = [20,10,50,24,12,56]
print(id(lista))
print()
# de mayor a menor --> Descendente
lista.sort(reverse=True)
print(lista)
# de menor a mayor --> Ascendente
lista.sort(reverse = False)
print(lista)
print()
print(id(lista))

```

sorted()

```python
lista = [20,10,50,24,12,56]
print(id(lista))
print()
# de mayor a menor --> Descendente
ordenada = sorted(lista, reverse=True)
print(ordenada)
# de menor a mayor --> Ascendente
print()
ordenada = sorted(lista)
print(ordenada)
print()
print(id(ordenada))

```
