DICCIONARIOS

```python
# definimos un diccionario con dict()
diccionario1 = dict(nombre = "Martin",
                    edad = 19, 
                    Estatura = 1.60)
print(diccionario1)
print(type(diccionario1))
```

```python
# diccionario vacío 
d1 = dict()
d2 = {}
print(type(d1))
print(type(d2))
```

```python
# Crear un diccionario y acceder a sus claves y valores
persona = {
    "Nombre": "Fernando Perez",
    "Edad": 30,
    "Profesion": "Ingeniero de Sofware",
    "Estatura": 1.65,
    "Genero": "Masculino"
}

# acceder a un diccionario 
print(persona["Nombre"])
print(persona["Edad"])
print(persona["Profesion"])

```

```python
# Crear un diccionario y acceder a sus claves y valores
persona = {
    "Nombre": "Fernando Perez",
    "Edad": 30,
    "Profesion": "Ingeniero de Sofware",
    "Estatura": 1.65,
    "Genero": "Masculino"
}

for clave, valor in persona.items():
    print(clave, " :", valor)

```
