DICCIONARIOS

```python
# definimos un diccionario con dict()
diccionario1 = dict(nombre = "Martin",
                    edad = 19, 
                    Estatura = 1.60)
print(diccionario1)
print(type(diccionario1))
```

```python
# diccionario vacío 
d1 = dict()
d2 = {}
print(type(d1))
print(type(d2))
```

```python
# Crear un diccionario y acceder a sus claves y valores
persona = {
    "Nombre": "Fernando Perez",
    "Edad": 30,
    "Profesion": "Ingeniero de Sofware",
    "Estatura": 1.65,
    "Genero": "Masculino"
}

# acceder a un diccionario 
print(persona["Nombre"])
print(persona["Edad"])
print(persona["Profesion"])

```

```python
# Crear un diccionario y acceder a sus claves y valores
persona = {
    "Nombre": "Fernando Perez",
    "Edad": 30,
    "Profesion": "Ingeniero de Sofware",
    "Estatura": 1.65,
    "Genero": "Masculino"
}

for clave, valor in persona.items():
    print(clave, " :", valor)

```

Método len()

```python
# Nos devuelve la cantidad de elementos de un diccionario

auto = {"Marca": "Toyota",
        "Cilindrada": "Automática",
        "Color": "Rojo",
        "Ruedas": 4
        }

# Aplicar len()
cantidad_elementos = len(auto)
print(cantidad_elementos)
```

VALUES()

```python
gato = {"Nombre": "Tom",
        "Edad": 4,
        "Genero": "Macho",
        "Raza": "Bombay"
        }

for valor in gato.values():
    print(valor)
```

KEYS()

```python
gato = {"Nombre": "Tom",
        "Edad": 4,
        "Genero": "Macho",
        "Raza": "Bombay"
        }

for valor in gato.keys():
    print(valor)
```

POP()

```python
libro = {"Titulo": "Bulevar",
         "Editorial": "Dragon"}

print(libro)
libro.pop("Editorial")
print(libro)
```

UPDATE()

```python
producto = {"Nombre": "Laptop",
            "Precio": 2600,
            "Categoria": "Tecnología"}

new_product = {"Stock": 100,
               "Garantia": 12,
               "Delivery": True}

producto.update(new_product)

for x,y in producto.items():
    print(x, " : ", y)
```
