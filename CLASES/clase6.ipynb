# Definir una clase 

```python
class Libro:
    def __init__ (self,codigo, titulo,autor, editorial, numero_paginas):
        self.codigo = codigo
        self.titulo = titulo
        self.autor = autor
        self.editorial = editorial
        self.numero_paginas = numero_paginas 
    
    
    def mostrar_informacion(self):
        cadena = f"Codigo: {self.codigo} -- Titulo: {self.titulo} -- Autor: {self.autor} -- Editorial: {self.editorial} -- Paginas: {self.numero_paginas}"
        print(cadena)



# programa principal 

libro1 = Libro(1234, "Calculo de una variable", "Stewar", "Dragon", 740)
#               cod         titulo                  autor   edit    n-p

print(libro1.titulo) # llamando a un atributo de la clase Libro 
libro1.mostrar_informacion() # llamando a un  método de la clase Libro


    
```

# __str__

```python
class Animal:
    def __init__(self, nombre, raza, edad):
        self.nombre = nombre
        self.raza = raza 
        self.edad = edad
        
    def __str__(self):
        return (f"Nombre: {self.nombre} -- "
                f"Raza: {self.raza} -- "
                f"Edad: {self.edad}")
    
    def mostrar_info(self):
        print(f"Nombre: {self.nombre}"
        f"Raza: {self.raza}"
        f"Edad: {self.edad}")
        

# programa principal 

lista_animales = []

animal1 = Animal("Toby", "Caniche", 10)
animal2 = Animal("Tom", " --- ", 8)
animal3 = Animal("Jerry", " --- ", 4)

lista_animales.append(animal1)
lista_animales.append(animal2)
lista_animales.append(animal3)

for a in lista_animales:
    print(a)

```

SISTEMA PARA MANEJO DE LIBROS  EN UNA BIBLIOTECA

```python
class Libro:
    def __init__(self, codigo, titulo, autor, editorial):
        self.__codigo = codigo
        self.__titulo = titulo
        self.__autor = autor
        self.__editorial = editorial
    
    @property
    def codigo (self):
        return self.__codigo
    
    @property
    def titulo (self):
        return self.__titulo
    
    @property
    def autor (self):
        return self.__autor
    
    @property
    def editorial (self):
        return self.__editorial
    
    def __str__(self):
        cadena = (f"Codigo: {self.codigo}\n" 
                f" Titulo: {self.titulo}\n"
                f"Autor: {self.autor}\n"
                f"Editorial : {self.editorial}\n")

class Biblioteca:
    def __init__(self):
        self.__lista_libros = []
    
    def tiene_numeros (self, texto): # hola
        for n in texto:
            if n.isdigit():
                return True
            return False
    
        
    def registrar_libro (self):
        while True:
            try:
                codigo = input("Ingrese un codigo para el libro (4 caracteres numericos): ")
                if len(codigo) == 4  and codigo.isdigit():
                    break 
                else:
                    print("El codigo debe tener 4 carateres numericos")
            except ValueError:
                print("Error en el ingreso de datos")
                    
        titulo = input("Ingrese el titulo: ")
        while True:
            try:
                autor = input("Ingrese el nombre del autor: ")
                if not self.tiene_numeros(autor):
                    break 
                else:
                    print("El nombre del autor no debe de tener numeros")
            except ValueError:
                print("Error en el ingreso de datos")
        
        editorial = input("Ingrese el nombre de la editorial: ")

        # crear el objeto libro
        libro = Libro(codigo, titulo, autor, editorial)
        
        # guardar el objeto en la lista
        self.__lista_libros.append(libro)
        print()
        print("Registro exitoso")
        print()

    
    def buscar_libro(self):
        while True:
            try:
                codigo = input("Ingrese un codigo para el libro a buscar (4 caracteres numericos): ")
                if len(codigo) == 4  and codigo.isdigit():
                    break 
                else:
                    print("El codigo debe tener 4 carateres numericos")
            except ValueError:
                print("Error en el ingreso de datos")


        for l in self.__lista_libros:
            if l.codigo == codigo:
                print("INFORMACIÓN DEL LIBRO BUSCADO: ")
                print()
                print(l)
    
    def eliminar_libro(self):
        while True:
            try:
                codigo = input("Ingrese un codigo para el libro a eliminar (4 caracteres numericos): ")
                if len(codigo) == 4  and codigo.isdigit():
                    break 
                else:
                    print("El codigo debe tener 4 carateres numericos")
            except ValueError:
                print("Error en el ingreso de datos")

        for l in self.__lista_libros:
            if l.codigo == codigo:
                self.__lista_libros.remove(l)
                print(f"LIBRO {l.titulo} ELIMINADO")
    
    def mostrar_libros (self):
        if len(self.__lista_libros) == 0:
            print("No existe registro de libros")
        else:
            libros = len(self.__lista_libros)
            print("--"*30)
            print("CANTIDAD DE LIBROS REGISTRADOS: ", libros)
            print("--"*30)
            for l in self.__lista_libros:
                print(l)
                print("--"*30)



def menu():
    print("MENU OPCIONES")
    print("1. Registrar libro")
    print("2. Buscar libro")
    print("3. Eliminar libro")
    print("4. Mostrar libros registrados")
    print("5. Salir")


def main():
    b = Biblioteca()
    while True:
        while True:
            try:
                menu()
                opcion = int(input("Ingrese una opcion: "))
                if 1<= opcion <= 5:
                    break 
                else:
                    print("La opcion no es valida")
            except ValueError:
                print("Error en el ingreso de datos")
        
        if opcion == 5:
            print("Saliendo del sistema .....")
            break
        
        if opcion == 1:
            b.registrar_libro()
        
        if opcion == 2:
            b.buscar_libro()
        
        if opcion == 3:
            b.eliminar_libro()
            
        if opcion == 4:
            b.mostrar_libros()

main()
        
    
        
```
