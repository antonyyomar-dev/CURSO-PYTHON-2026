MATCH 

```python
num1 = int(input("Ingrese un primer numero: "))
num2 = int(input("Ingrese un segundo numero: "))
operacion = input("Ingrese tipo de operación (** * / + - ): ")
resultado = 0

match operacion:
    case "**":
        resultado = num1 ** num2
    case "*":
        resultado = num1 * num2
    case "/":
        resultado = num1 / num2
    case "+":
        resultado = num1 + num2
    case "-":
        resultado = num1 - num2
    case _:
        print("El tipo de operacion no es correcto")

# print("La operacion ", operacion , 
      # " de [ ", num1 ,  " + ", num2, "] es = ", resultado)

print(num1, " ", operacion, " ", num2,  " = ", resultado)
```

Bucle while 

```python
while True:
    try:
        nota = float(input("Ingrese una nota: "))
        if 0<= nota <= 20:
            break 
        else:
            print("La nota debe de estar entre 0 y 20")
    except ValueError:
        print("Dato invalido")


if 0<= nota < 13:
    print("Estas desaprobado")
    
if 13<= nota <= 20:
    print("Estas aprobado")
    
```

Funciones

```python
# una función es un bloque de código que se ejecuta cuando se llama o invoca
# def evaluar_nota_del_alumno():

def evaluarNotaDelAlumno(nota):
    while True:
        try:
            if 0<= nota <= 20:
                break 
            else:
                print("La nota debe de estar entre 0 y 20")
        except ValueError:
            print("Dato invalido")


    if 0<= nota < 13:
        print("Estas desaprobado")
        
    if 13<= nota <= 20:
        print("Estas aprobado")

nota = float(input("Ingrese una nota: "))
evaluarNotaDelAlumno(nota)

```
